Vanishing Grad- Here, as we move down to the lower layers, the gradient keeps getting smaller and smaller. This leads to lower layers training slower than higher layers.<br>
If we're using sigmoid, the function saturates to 1 for higher values of input. The gradient is approx. 0 for this region. This little gradient is then passed down to the lower layers and it's lessened even more till it's virtually non-existent.

Exploding Gradients- This occurs more in Deep NNs. Here, the gradient explodes as we move down the layers due to large difference b/w weights and loss calculated.


In order to combat this, we have to ensure that the input variance to each neuron is equal to the output variance to each neuron. This is not possible in practice for most cases, but there is a good estimation for it. 

fan-in is the no. of inputs and fan-out is the no. of neurons of that layer. For each neuron we take fan(avg) and then utilize it in the xavier initialization.
