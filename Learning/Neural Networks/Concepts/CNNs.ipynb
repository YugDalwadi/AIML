<h1>Convolution</h1>
In a convolution, we calculate the dot of one function with another and compute their integral

Padding- Zeroes added to facilitate for receptive field of higher layer.<br>
Stride- spacing b/w two diff. receptive fields.<br>
fh,fw- field width and height. It is span of field in either dirn. <br>

Neuron at i,j in higher layer is connected to neurons spanning from i\*sh to i\*sh+fh-1 and j\*sw to j\*sw+fw-1 
note that size of receptive field is same, but distancing b/w two fields is changed to sh,sw instead of 1,1.

Filter- A filter is the set of weights which will be assigned to the neurons of the higher layer. This higher layer is a feature map.<br>
A layer full of neurons using the same filter is known as a feature map.

So, a Convolutional Layer has multiple filters and outputs one feature map per filter. Hence, it is more accurately represented in 3D. See Pg. 452 for better understanding.

In the case of a colored image, Each map will look at a 3D cube of a certain block of pixels.

A convolutional Layer is made up of multiple convolutional maps. Each feature map is connected to the map below it. The lowermost map is connected to the prev. layer's uppermost map.

<h1>POOLING LAYERS</h1>

The goal of pooling layers is to shrink the input image in order to reduce computational load. Pooling layers dont have weights of their own. They just have a receptive field on which the compute the aggregate function. The function may be anything like max or mean. 

Max pooling layers are the most common.

Pooling layers also introduce some level of invariance to small translations to the images. Check Pg.458 for examples.<br>
However, invariance isn't always good. In semantic segmentation, we classify each pixel as belonging to a certain object. If an object shifts by one pixel, then we don't want invariance. We want equivariance, where a small change in input should lead to a small change in output.

Max Pooling offers stronger invariance and is easier to compute as compared to Average Pooling.

A common mistake is using too large kernels. It is better to stack 2 3x3 layers instead of using 1 5x5 layer. It requires less computational load. An exception is the first conv. layer. It can have a larger kernel since this will reduce the dims of image without losing too much info.
