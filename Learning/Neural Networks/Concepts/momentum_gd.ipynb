<h1>Momentum Optimization</h1>

A ball rolling on a slope will start out slow and eventually reach Vt. This is the logic behind momentum optimization. Simple GD will take small and regular steps to the bottom which will increase training time.

Beta is the momentum. It acts as a sort of frictional coeff. in order to prevent m from growing too large and cause GD to fail. 0 is high frn, 1 is no frn. It is typically .9

Here, m is the step size. It is a linear combination of prev step size along with new gradient step. This makes it less prone to forces that may suddenly change its dirn.

```python
X = np.random.rand(200,1)
y = 4 +3*X + np.random.randn(200,1) 
X = np.c_[np.ones((200,1)),X] 
```

```python
class MGD():
    def __init__(self,X,y,n_epoch,eta,beta):
        self.X = X
        self.y = y
        self.n_epoch = n_epoch
        self.eta = eta
        self.beta = beta
    
    def fit(self,X,y=None):
        return self

    def transform(self):
        theta = np.random.randn(self.X.shape[1],1)
        mom = np.zeros(theta.shape)
        for i in range(self.n_epoch):
            grad = ((np.matmul(self.X,theta)-self.y).T.dot(self.X)).T
            mom = self.beta*mom - self.eta*grad
            theta = theta+mom
        return theta
```

```python
mgd = MGD(X,y,100,eta=0.01,beta=0.9)
theta = mgd.transform()
theta
```
