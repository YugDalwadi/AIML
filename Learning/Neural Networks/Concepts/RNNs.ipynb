<h1>RNNs(Recurrent Neural Networks)</h1>

At each time step =t(called a frame), each neuron receives the current input x(t)
and some hidden state h_t = f(h_(t-1) , x_t).
The output at t=t will be y_t. It is computed by a weighted sum of x_t part and h_(t-1) part.

<h1> Seq2Seq</h1>
Here, a seq. of inputs X vec. is simultaneously given to the RNN. It produces the output as a sequence of Y.<br>
This type of RNN is used for Time Series prediction, where we give input of past N days to N RNN neurons. The final neuron gives the output for N+1'th day.<br>

<h1>Seq2Vec</h1>
Here, a seq. of inputs X is given simultaneously to the RNN. In the end, we take the Y vector of final recurrent neuron. This is used for senti analysis, where we feed the network word embeddings corresponding to a movie review, and the network would output a senti score.
