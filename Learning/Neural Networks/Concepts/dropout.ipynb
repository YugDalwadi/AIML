<h1>DROPOUT</h1>

In dropout, at each epoch we take some probability 'p'(dropout rate) of a neuron being droppped out. All such neurons are dropped out(HL and IL) and then model is trained on the ones that are left. This helps in creating robust connections between the remaining neurons. The dropepd out neurons are then returned and the process is repeated once again.<br>
If there are 10 neurons, there are 2^10=1024 diff. possible neural nets. So, it's unlikely that the same neural net will be chosen for all epochs. On an average, we'll train 1024 different neural networks. Neurons also become less sensitive to slight input changes in the neurons they're connected to. It's like an ensemble of 1024 different neural networks.

p is usually 20-30% in RNNs and 40-50% in CNNs.

There is an important detail with neural nets trained using dropout techniques. Let p=0.5. In this case, on average, each neuron will be connected to half the no. of neurons it was connected to in the original network. To compensate, we multiply the input connection weights by 0.5 after training. If we don't, each neuron will get signals roughly twice as large as to what it was trained on.

MORE GENERALLY, we need to multiply each weight by (1-p) i.e the keep probability in order to compensate for the signals. OR we can divide each neuron's output by 1-p to get almost the same result.<br>
For intuition, if p=0.4 neurons are being dropped out, then the resulting neural net experiences 40% less neurons. So, it effectively experiences 60% neurons. So, we must divide by 60% to get the original back(0.6 0.6 cuts out)

Dropout tends to slow down convergence, but it is well worth it after the model is tuned properly.

```python
from sklearn.datasets import fetch_california_housing
from sklearn.preprocessing import StandardScaler
from sklearn.model_selection import train_test_split

housing = fetch_california_housing()

xTrain,xTest,yTrain,yTest = train_test_split(housing.data, housing.target)
xTrain,xVal,yTrain,yVal = train_test_split(xTrain,yTrain)
```

```python
scaler = StandardScaler()
xTrain = scaler.fit_transform(xTrain)
xTest = scaler.transform(xTest)
xVal = scaler.transform(xVal)
```

```python
input_ = keras.layers.Input(shape=xTrain.shape[1:])
dl1 = keras.layers.Dropout(rate=0.2)(input_)
hl1 = keras.layers.Dense(30,activation="relu")(dl1)
dl2 = keras.layers.Dropout(rate=0.2)(hl1)
hl2 = keras.layers.Dense(30,activation="relu")(dl2)
dl3 = keras.layers.Dropout(rate=0.2)(hl2)
output_ = keras.layers.Dense(1)(dl3)

model = keras.Model(inputs=[input_], outputs=[output_])
model.compile(loss="mse", optimizer="adam")
```

```python
model.fit(xTrain,yTrain,validation_data=(xVal,yVal) ,epochs=40)
```

```python
model.evaluate(xTest,yTest)
```

```python
input_ = keras.layers.Input(shape=xTrain.shape[1:])
hl1 = keras.layers.Dense(30,activation="relu")(input_)
hl2 = keras.layers.Dense(30,activation="relu")(hl1)
output_ = keras.layers.Dense(1)(hl2)

model = keras.Model(inputs=[input_], outputs=[output_])
model.compile(loss="mse", optimizer="adam")
```

```python
model.fit(xTrain,yTrain,validation_data=(xVal,yVal) ,epochs=20)
```

```python
model.evaluate(xTest,yTest)
```
