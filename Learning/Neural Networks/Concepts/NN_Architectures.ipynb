<h1>LeNet-5</h1>
Pg-463

Widely used for MNIST.

Only the 1st conv layer uses padding. Rest have 0 padding, which is why the image size shrinks as we go deeper down.

The pooling layers are slightly complex. They have a learnable bias term and a coefficient. Each neuron computes the mean and multiplies by the coeff. and then adds the bias. Each pooling map has a single bias and coeff.<br>
Conv. neurons of C3 aren't connected to all pool layers of S2. This is to reduce computational load and force a break of symmetry.

The output layer is fully connected to prev. fully connected layer. However, it doesn't compute a matrix multiplication of input vector and weight vector. Instead, each neuron computes the square of the euclidean distance b/w it's input vector and output weights. The further away the input is from the parameter vector, the larger the RBF(Value we compute) is.
