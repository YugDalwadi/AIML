Many datasets have thousands or more features. Training on such data is very slow due to the amount of features. In such cases, in order to speed up training(Or even improve the performance of models by driving out noise, such as the Dec Tree 45deg rotation case), we use Dim Reduction methods. They can also be used for data visualization by reducing dims down to 2D or 3D and visually detecting things such as clusters, anomalies, etc.

Dim Reduction in general will give slightly worser performing models, but may speed up training by a lot.

<h1>THE CURSE OF DIMENSIONALIY</h1>

In a unit sq, there is a 0.4% chance of a random point being close to the border, however in 10000D unit hypercube, this becomes 99.999999%. In higher dims, extremes are much more likely than "normal" datapoints.<br>
Similarly, if we take 2 random points on a unit sq, their avg. distance b/w each other will be 0.52, however for a 1000000D hypercube, it will be approx. 408.25. Clearly, this will result in a very sparse dataset and new instances will be far away too. Hence, the datasets are prone to overfitting.

<h1>PROJECTION</h1>

In most real world problems, the bulk of the correlation b/w features and labels/classes is restricted to a few features. Many other features contribute too less to the correlation and can be safely ignored.<br>

<h1>MANIFOLD LEARNING</h1>

Swiss Roll is an example of a 2D Manifold. A 2D Manifold is a 2D object that can be bent in 3D to form a 3D figure.

Many Dim-Redn algos work by assuming that the training instances all lie on a manifold. This is the Manifold Hypothesis. It holds that most real world problems lie close to a much lower dimension manifold.

Manifold Hypothesis usually also comes with another assumption, i.e that the given reg/class task at hand will be simpler in a lower dimension. However, this is not always true. It depends on the decision boundary in the higher dimension. See PG 219 for a good example of this on the swiss roll dataset.
