```python
import numpy as np
import pandas as pd
from matplotlib import pyplot as plt
```

```python
from sklearn.datasets import make_moons
X,y = make_moons(n_samples=10000, noise=0.4) # moons is a toy dataset for classification

plt.scatter(X[:,0], X[:,1], c=y)
plt.show()
```

```python
from sklearn.model_selection import train_test_split
from sklearn.tree import DecisionTreeClassifier

xTrain, xTest, yTrain,yTest = train_test_split(X,y)
decTree = DecisionTreeClassifier()
```

```python
from sklearn.model_selection import GridSearchCV

searcher = GridSearchCV(decTree,
                        param_grid=
                        {
                            "max_depth":[1,2,3,4], "max_leaf_nodes":[2,3,4,5,6]
                        }
                        )
searcher.fit(xTrain,yTrain)
```

```python
from sklearn.metrics import accuracy_score

score = searcher.best_score_
preds = searcher.predict(xTest)
acc = accuracy_score(yTest,preds)
print(score,acc)
```
