```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
dataset = pd.read_csv("iris.csv")
X = dataset.drop("species",axis=1)
X = X.iloc[:,2:]
y = dataset["species"]
```

Dec Trees can be used for both classification and regression, as well as even multioutput tasks.<br>
They require almost no data prep and can be trained with unscaled and uncentred data 

A node's sample attr. counts how many training instances it applies to. The value attr gives the exact number of samples for each class possible

At depth=0, we have 50 of all 3 classses.<br>
At d=1, we have 50 on left leaf and 50,50 on right branch.<br>
At d=2, we have total 49+5 + 1+45=100 on the two leaves. This is due to 50 already being classified as setosa on the d=1 leftmost leaf.

Sklearn used CART for dec trees, which only produces binary trees. Algos such as ID3 can produce >2 children also.

```python
from sklearn.tree import DecisionTreeClassifier, export_graphviz

dec_clf = DecisionTreeClassifier(max_depth=2)
dec_clf.fit(X,y)
export_graphviz(
                dec_clf,
                out_file="irisTree.dot",
                feature_names=X.columns,
                class_names=y.unique(),
                rounded=True,filled=True
                )
```

GINI IMPURITY

It indicates how pure the node is in terms of classes.
If a node has only 1 class, it's impurity is 0.
For the leftmost d=2 node, it's impurity is 1-(0/54)^2 -(49/54)^2 -(5/54)^2 ~ 0.168

CLASS PROBABILITIES

CART Algo
PG 179 FOR EQNS

Sklearn uses CART(Classification And Regression Tree) algo to train dec trees. The algo first splits the training set according to a particular feature k and threshold Tk. It searches greedily for a good enough split. 'Goodness' of split is calculated by the cost fn given. It basically tries to minimize the impurity of each side of split weighted according to the no. of instances in each split.

Once an appropriate split is found, it then goes down and repeats the same process for the node's children until either the impurity tolerance is reached or the max depth is reached.

A Dec Tree is a non-parametric algo. It doesn't have a fixed no. of parameters from the start.(Eg- in linear regression, poly.reg,etc)<br>
So, an unregularized algo sticks very well to the training set and overfits very easily. To reduce this, we have some hyperparameters we can tune-<br>
1) max_depth
2) min_samples_split(min. no of samples a node must have b4 it can split)
3) min_samples_leaf(min. no of samples a leaf node must have)
4) max_leaf_nodes(max no. of leaf nodes)
5) max_features(Max no. of features that are evaluated for splitting at each node.)
6) min_weight_fraction_leaf(sames as min_samples_leaf but expressed as fraction of no. of weighted instances.)

The main issues with Dec Trees are-
1) V. prone to overfitting
2) Scikit's training algo is stochastic, so on the same training data we may have vastly different splits
3) The model is very sensitive to small variations in dataset
