```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib
```

```python
X = np.random.rand(200,1) #rand gives uniform distribution, it is constant conc of dist. across the interval
y = 4 +3*X + np.random.randn(200,1) ## Random normal dist(bell curve)
X = np.c_[np.ones((200,1)),X] # we do this so that bias term can be added as 1*bias = bias only
```

```python
plt.scatter(X[:,1],y)
plt.show()
```

```python
theta = np.random.randn(2,1)
def gradientDescent(X,y,eta,n_iter,m,theta):
    theta = theta
    for i in range(n_iter):
        gradient = (2/m)*(X.T).dot((np.matmul(X,theta)-y))
        theta = theta - eta*gradient
    return theta

theta = gradientDescent(X,y,eta=0.1,n_iter=1000,theta=theta,m=200)
print(theta)
```
