SGD selects a random data point and computes MSE on it at each step. This is why its much faster than BGD for large datasets.

This random behavior may help the algo get out of a local minima

We will implement a learning schedule-- This slowly decreases learning rate in order to decrease random jumps around the minima.



```python
X = np.random.rand(200,1)
y = 4 +3*X + np.random.randn(200,1) ## Random normal dist
X = np.c_[np.ones((200,1)),X]
len(X)
```

```python

t0,t1=5,50

def learning_schedule(t):
    return t0/(t1+t)

def SGD(X,y,n_epoch):
    theta = np.random.rand(2,1) 
    m = len(X)
    for epoch in range(n_epoch):
        for i in range(m):
            randIndex = np.random.randint(m) # excludes m
            randX = X[randIndex:randIndex+1]
            randY = y[randIndex:randIndex+1]
            
            gradient = 2*(randX.T).dot((np.matmul(randX,theta)-randY)) ## 2 instead of 2/m since only 1 element now, so m=1
            eta = learning_schedule(epoch*m + i)
            theta = theta - eta*gradient
    return theta

theta = SGD(X,y,50)
theta
```
