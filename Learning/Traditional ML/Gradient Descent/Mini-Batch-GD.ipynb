```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
X = np.random.rand(2000,1) # 200 samples
y = 4 +3*X + np.random.randn(2000,1) ## Random normal dist
X = np.c_[np.ones((2000,1)),X]
```

```python

a,b=5,50
def learning_schedule(t):
    return a/(b+t) # at  t=0, eta = a/b
X.shape
batchSize = 100
def MBGD(X,y,n_epochs,batchSize):
    theta = np.random.rand(2,1)
    
    for i in range(n_epochs):
        rand_ix = np.random.randint(0,len(X)-batchSize)
        batchX = X[rand_ix:rand_ix+batchSize]
        batchY = y[rand_ix:rand_ix+batchSize]
        print((np.matmul(batchX,theta)-batchY).T.shape,batchX.shape)
        grad = (2/batchSize)*(np.matmul(batchX,theta)-batchY).T.dot(batchX)
        eta = learning_schedule(i/2)
        theta = theta-eta*grad.T
    return theta

theta = MBGD(X,y,5,100)
theta
```
