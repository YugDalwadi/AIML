<h1>BOOSTING</h1>

Boosting is any ensemble method that combines several weak learners(models that perform slightly better than random choice models) and train predictors sequentially(in order to correct it's predecessor) to create a strong learner.

<h1>ADABOOST(Adaptive Boosting)</h1>
For eqns, go to Pg.201,202

In this method, the instances that are labelled wrongly are given more weight as compared to instances that are labelled correctly.

Basically, Initially each instance has an equal likelihood of being picked. i.e, 1/m probability.
For the 2nd iteration, the instances that were misclassified have a higher probability of being picked.

In detail-<br>
The 1st iter will be standard since each instance has equal probability of being picked.<br>
The 2nd iter will have a weight multiplied ONLY to instances that were misclassified in the 1st iteration. This leads to a higher probability of picking those instances.



```python
from sklearn.datasets import make_moons
X,y=make_moons(10000, noise=0.4)

xTrain = X[int(len(X)*0.8):]
yTrain = y[int(len(X)*0.8):]

xTest = X[:int(len(X)*0.8)]
yTest = y[:int(len(X)*0.8)]
```

```python
plt.scatter(xTrain[:,0], xTrain[:,1],c=[yTrain], alpha=0.5)
plt.show()
```

```python
from sklearn.ensemble import AdaBoostClassifier
from sklearn.tree import DecisionTreeClassifier

decTree = DecisionTreeClassifier(max_depth=1)
adaClf = AdaBoostClassifier(decTree,n_estimators=200,learning_rate=0.5)
adaClf.fit(xTrain,yTrain)

score = adaClf.score(xTest,yTest)
print(score)
```
