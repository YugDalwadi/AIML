An ensemble is a collection of many models which you've trained to obtain good accuracy on the test set.

1) Voting Classifiers
These are for classifiers and they can be of two types- <br>
Hard Voting- Majority vote <br> 
Soft Voting- Averages the probabilities of all classifiers and votes acc. to that accuracy.

<h1>BAGGING AND PASTING</h1> <br>
We can train an ensemble in two ways- By using a diverse set of classifiers, or by using the same algo but on different random subsets of the training set. <br>
In the latter, when sampling is performed with replacement, it's called bagging. If it's done without replacement, it's called pasting.

Bootstrapping- Resampling with replacement<br>

Each individual predictor will have higher bias than a single predictor trained on entire data, but the aggregator will have a lower bias and variance.<br>
Generally, bias is similar to that of a model trained on full dataset, but variance is lower.

BAGGING > PASTING usually

Out OF Bag (OOB)- Let's say we have a bagging classifier with bootstrap=True. Each classifier instance will not train on the entire data, but it will train on approx. 63% at once. Rest 37% instances are called OOB instances for that classifier. Sklearn has an option to show evaluations of OOB Scores on individual models and average it out to get a rough evaluation of the ensemble.
